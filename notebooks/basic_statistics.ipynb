
# Estadística básica: análisis de transacciones

## Pregunta de investigación
¿Cómo varían los importes de las transacciones según el canal?

## Fuente de datos
Conjunto de datos sintético generado con NumPy.

## Variables
- channel: canal de la transacción (online o en tienda).
- amount: importe de la transacción en unidades sintéticas.

## Preparación de los datos
El conjunto de datos contiene 200 observaciones. Se ha introducido intencionadamente un valor faltante en «amount», que se ha excluido del análisis estadístico.

## Política sobre valores perdidos
Las filas con valores perdidos en «amount» o «channel» se excluyen mediante la función `dropna()`.

## Análisis
Se utilizan estadísticas descriptivas y visualizaciones para comparar los importes de las transacciones entre los distintos canales.

```python
import numpy as np
import pandas as pd
import plotly
import plotly.express as px

rng = np.random.default_rng(42)

df = pd.DataFrame({
    "channel": ["Online"] * 100 + ["Store"] * 100,
    "amount": np.round(rng.lognormal(3.5, 0.6, 200), 2)
})

df.loc[0, "amount"] = np.nan
display(df)
```

```python
print("Versions:")
print("NumPy:", np.__version__)
print("Pandas:", pd.__version__)
print("Plotly:", plotly.__version__)

print("\nTotal rows:", len(df))

print("\nMissing values:")
print(df[["amount", "channel"]].isna().sum())
```

```python
clean = df.dropna(
    subset=["amount", "channel"]
).copy()

if len(clean) < 2 or not np.isfinite(clean["amount"]).all():
    raise ValueError("Need at least two finite amounts")

print("Analyzed rows:", len(clean))
```

```python
print("Descriptive Statistics:")
print(clean["amount"].describe().round(2))

print("\nMedian:")
print(round(clean["amount"].median(), 2))

print("\nStatistics by channel:")
print(
    clean.groupby("channel")["amount"].agg(
        ["count", "mean", "median"]
    ).round(2)
)
```

```python
histogram = px.histogram(
    clean,
    x="amount",
    nbins=20,
    title="Transaction Amount Distribution",
    labels={
        "amount": "Amount (synthetic units)"
    }
)

histogram.show()
```

```python
box = px.box(
    clean,
    x="channel",
    y="amount",
    title="Transaction Amounts by Channel",
    labels={
        "amount": "Amount (synthetic units)",
        "channel": "Channel"
    }
)

box.show()
```
