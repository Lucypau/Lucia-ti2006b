## Pregunta de investigación
¿Cuál es el promedio de cantidad de transacciones fraudulentas y legitimas?

## Procedencia de los datos
El conjunto de datos se ha creado con fines educativos. No contiene información real.

## Resultado esperado
Un resumen de los importes de las transacciones y del número de transacciones legítimas y potencialmente fraudulentas.

```python
import sys
import pandas as pd

print("Python version:", sys.version)
print("Pandas version:", pd.__version__)
```

```python
datos = {
    "t_id": [1, 2, 3, 4, 5, 6],
    "cantidad": [150, 2500, 80, 4500, 300, 1200],
    "Estatus": [ "Legitimo","Fraude","Legitimo","Fraude", "Legitimo", "Legitimo"]
}

df = pd.DataFrame(datos)

df
```

```python
print("Resumen:")
print(df.groupby("Estatus")["t_id"].count())

print("\nPromedio de cantidad:")
print(df.groupby("Estatus")["cantidad"].mean())

print("\nTransacciones totales:", len(df))
```

## Conclusión

El análisis muestra que los datos contienen 4 transacciones legítimas y 2 transacciones fraudulentas. En donde el promedio de cantidad de transacciones legítimas es de 432.5 y el promedio de fraude es de 3500. Demostrando que las transacciones fraudulentas son de mayor cantidad.
